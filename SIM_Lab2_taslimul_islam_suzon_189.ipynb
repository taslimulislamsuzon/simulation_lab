```python
import numpy as np

student_id = "189"
d1 = int(student_id[-1])
d2 = int(student_id[-2])
A = np.array([
    [d1 + 2, d2 + 1],
    [2 * d1, d2 + 2]
])

print(f"d1 = {d1}")
print(f"d2 = {d2}")
print("\nMatrix A:")
print(A)
```

```python
import numpy as np
shape = A.shape
print(f"Shape of A: {shape}")

det = np.linalg.det(A)
print(f"Determinant of A: {det:.4f}")

rank = np.linalg.matrix_rank(A)
print(f"Rank of A: {rank}")

eigenvalues, eigenvectors = np.linalg.eig(A)
print(f"Eigenvalues of A: {eigenvalues}")

if not np.isclose(det, 0):
    A_inv = np.linalg.inv(A)
    print("\nInverse of A:")
    print(A_inv)
else:
    print("\nMatrix A is singular and does not have an inverse.")
```

```python
import numpy as np

B = A.copy()
B[0, 0] += 1

print("Original Matrix A:")
print(A)
print("\nModified Matrix B (Added +1 to element [0,0]):")
print(B)
```

```python
shape_B = B.shape
print(f"Shape of B: {shape_B}")

det_B = np.linalg.det(B)
print(f"Determinant of B: {det_B:.4f}")

rank_B = np.linalg.matrix_rank(B)
print(f"Rank of B: {rank_B}")

eigenvalues_B, eigenvectors_B = np.linalg.eig(B)
print(f"Eigenvalues of B: {eigenvalues_B}")

if not np.isclose(det_B, 0):
    B_inv = np.linalg.inv(B)
    print("\nInverse of B:")
    print(B_inv)
else:
    print("\nMatrix B is singular and does not have an inverse.")
```
