```python
import numpy as np
scalar = 3.7
arr = np.array([0.0, 0.5, 1.0, 1.5, 2.0, 2.5])

print("Scalar:", scalar)
print("Array:", arr)

print("Sin:", np.sin(arr))
print("Cos:", np.cos(arr))
print("Tan:", np.tan(arr))
print("Sinh:", np.sinh(arr))
print("Cosh:", np.cosh(arr))
print("Tanh:", np.tanh(arr))
```

```python
matrix_3x2 = np.array([[-0.5, 0.8],
                       [1.5, -2.3],
                       [0.0, 3.1]])

print("Original 3x2 Matrix:\n", matrix_3x2)

clipped_matrix = np.clip(matrix_3x2, -1.0, 1.0)
print("\nArcsin (Clipped):\n", np.arcsin(clipped_matrix))
print("Arccos (Clipped):\n", np.arccos(clipped_matrix))
print("\nExponential (e^x):\n", np.exp(matrix_3x2))
print("Logarithm (ln|x| + epsilon):\n", np.log(np.abs(matrix_3x2) + 1e-9))
print("\nAbsolute Value:\n", np.abs(matrix_3x2))
print("Square Root of Absolute Values:\n", np.sqrt(np.abs(matrix_3x2)))
print("\nRemainder when divided by 2:\n", np.remainder(matrix_3x2, 2))
print("\nRound:\n", np.round(matrix_3x2))
print("Floor:\n", np.floor(matrix_3x2))
print("Ceil:\n", np.ceil(matrix_3x2))
```

```python
matrix_2x3 = np.array([[2.1, -4.7, 5.5],
                       [0.3, 8.9, -1.2]])

print("Original 2x3 Matrix:\n", matrix_2x3)

max_val = np.max(matrix_2x3)
min_val = np.min(matrix_2x3)
max_idx = np.unravel_index(np.argmax(matrix_2x3), matrix_2x3.shape)
min_idx = np.unravel_index(np.argmin(matrix_2x3), matrix_2x3.shape)

print(f"\nMax value: {max_val} at index {max_idx}")
print(f"Min value: {min_val} at index {min_idx}")

vec_len = np.linalg.norm(matrix_2x3)
print(f"Vector length (L2 Norm): {vec_len:.4f}")

row_sorted_desc = np.sort(matrix_2x3, axis=1)[:, ::-1]
flat_sorted_desc = np.sort(matrix_2x3.flatten())[::-1]

print("\nRow-wise sorted (descending):\n", row_sorted_desc)
print("Entire matrix sorted (descending):\n", flat_sorted_desc)
print(f"\nSum: {np.sum(matrix_2x3):.4f}")
print(f"Product: {np.prod(matrix_2x3):.4f}")
print(f"Median: {np.median(matrix_2x3):.4f}")
print(f"Mean: {np.mean(matrix_2x3):.4f}")
print(f"Standard Deviation: {np.std(matrix_2x3):.4f}")
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(6, 4))
sns.heatmap(matrix_2x3, annot=True, cmap="coolwarm", fmt=".2f", cbar=True)
plt.title("Heatmap of 2x3 Matrix")
plt.show()

random_uniform_matrix = np.random.uniform(low=-10.0, high=10.0, size=(2, 3))
print("\nRandom Uniform 2x3 Matrix:\n", random_uniform_matrix)
```
