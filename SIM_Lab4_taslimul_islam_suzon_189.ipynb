```python
import numpy as np
import scipy.stats as stats
data1 = np.array([18.4, 19.1, 17.9, 18.7, 18.3])
data2 = np.array([20.2, 20.5, 20.1, 20.3, 20.4])
```

```python
mean1 = np.mean(data1)
mean2 = np.mean(data2)
var1 = np.var(data1, ddof=1)
var2 = np.var(data2, ddof=1)
mean_diff = mean1 - mean2

print(f"Data 1 Mean: {mean1:.4f}, Variance (ddof=1): {var1:.4f}")
print(f"Data 2 Mean: {mean2:.4f}, Variance (ddof=1): {var2:.4f}")
print(f"Mean Difference (Data 1 - Data 2): {mean_diff:.4f}\n")
```

```python
n1 = len(data1)
n2 = len(data2)
df = n1 + n2 - 2

pooled_var = ((n1 - 1) * var1 + (n2 - 1) * var2) / df
se = np.sqrt(pooled_var * (1/n1 + 1/n2))
t_stat = mean_diff / se
p_value = 2 * (1 - stats.t.cdf(abs(t_stat), df=df))

print(f"t-statistic: {t_stat:.4f}")
print(f"Degrees of Freedom (df): {df}")
print(f"Two-tailed p-value: {p_value:.6f}\n")
```

```python
alpha = 0.05
t_critical = stats.t.ppf(1 - alpha/2, df=df)
margin_of_error = t_critical * se
ci_lower = mean_diff - margin_of_error
ci_upper = mean_diff + margin_of_error

print(f"95% Confidence Interval for the mean difference: ({ci_lower:.4f}, {ci_upper:.4f})\n")
```
